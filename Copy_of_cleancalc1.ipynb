```python
import collections
import fractions

# Define base rates for different world types (resources/min for a level 1 extractor)
# Using fractions for exact arithmetic
default_world_base_rate = fractions.Fraction(75, 10) # 7.5
gen2_world_base_rate = fractions.Fraction(150, 10) # 15.0
uranium_default_base_rate = fractions.Fraction(100, 10) # 10.0
uranium_gen2_base_rate = fractions.Fraction(500, 10) # 50.0

# Define boost percentages as fractions, now dependent on world type
boost_percentages = {
    'nuclear': {
        'default': fractions.Fraction(40, 100),  # +40% for default worlds
        'gen2': fractions.Fraction(60, 100)      # +60% for gen2 worlds
    },
    'coal': fractions.Fraction(20, 100) # +20% for all worlds
}

# Define extractor level multipliers - these are exact
extractor_multipliers = {
    'default': {
        1: fractions.Fraction(1),
        2: fractions.Fraction(3, 2), # 1.5
        3: fractions.Fraction(2),
        4: fractions.Fraction(3),
        5: fractions.Fraction(4)
    },
    'gen2': {
        1: fractions.Fraction(1),
        2: fractions.Fraction(2),
        3: fractions.Fraction(4),
        4: fractions.Fraction(8),
        5: fractions.Fraction(10)
    }
}

# Define building tier multipliers as fractions, using integer keys
building_tier_multipliers = {
    1: fractions.Fraction(1),   # 100%
    2: fractions.Fraction(3, 2), # 150%
    3: fractions.Fraction(2),   # 200%
    4: fractions.Fraction(3)    # 300%
}

# Define the assumed building tiers for the optimization problem
# All building types except 'Earth Teleporter' default to '300%', 'Earth Teleporter' to '100%'
assumed_building_tiers = {
    "Furnace": 4, # Corresponds to 300%
    "Workshop": 4, # Corresponds to 300%
    "Forge": 4, # Corresponds to 300%
    "Machine Shop": 4, # Corresponds to 300%
    "Industrial Factory": 4, # Corresponds to 300%
    "Manufacturer": 4, # Corresponds to 300%
    "Earth Teleporter": 1 # Corresponds to 100%
}

# List of all resources, separating Uranium Ore due to its unique rates
standard_resources = ["Wood Log", "Coal", "Iron Ore", "Copper Ore", "Stone", "Wolframite"]
uranium_resource = "Uranium Ore"

# Initialize a dictionary to store resource production rates
resource_production_rates = {}
```

```python
import math
import fractions

# --- Calculate Total Resource Production ---

for resource in standard_resources + [uranium_resource]:
    total_production = fractions.Fraction(0)
    node_count = node_counts[resource]

    # Determine the base rate for the resource based on world type and resource type
    if resource == uranium_resource:
        base_rate = uranium_default_base_rate if world_type == 'default' else uranium_gen2_base_rate
        extractor_rate = base_rate # Uranium extractors do not level up, so no multiplier
    else:
        base_rate = default_world_base_rate if world_type == 'default' else gen2_world_base_rate
        # Apply extractor multiplier to standard resources
        extractor_rate = base_rate * current_extractor_multiplier

    num_nuclear_boosted = fractions.Fraction(nuclear_boosted_counts[resource])
    num_coal_boosted = fractions.Fraction(coal_boosted_counts[resource])
    num_unboosted = fractions.Fraction(node_count) - num_nuclear_boosted - num_coal_boosted

    # Get the correct boost factors from the global dictionary
    nuclear_boost_factor = boost_percentages['nuclear'][world_type]
    coal_boost_factor = boost_percentages['coal']

    production_nuclear_boosted = num_nuclear_boosted * extractor_rate * (fractions.Fraction(1) + nuclear_boost_factor)
    production_coal_boosted = num_coal_boosted * extractor_rate * (fractions.Fraction(1) + coal_boost_factor)
    production_unboosted = num_unboosted * extractor_rate

    total_production = production_nuclear_boosted + production_coal_boosted + production_unboosted

    resource_production_rates[resource] = total_production

# You can now access the calculated rates via the resource_production_rates dictionary
```

```python
def calculate_min_boosted_fractional(target_production, total_nodes_for_res, resource_name, world_type_val, extractor_multiplier_val):
    """
    Calculates the minimum fractional boosted nodes required for a given resource
    to meet a target production rate, considering world type and extractor multiplier.
    """
    if resource_name == uranium_resource:
        base_rate = uranium_default_base_rate if world_type_val == 'default' else uranium_gen2_base_rate
        R_base_per_node = base_rate # Uranium extractors do not level up, so no multiplier
    else:
        base_rate = default_world_base_rate if world_type_val == 'default' else gen2_world_base_rate
        R_base_per_node = base_rate * extractor_multiplier_val

    nuclear_boost_factor_current_world = boost_percentages['nuclear'][world_type_val]

    denominator = R_base_per_node * nuclear_boost_factor_current_world
    if denominator != fractions.Fraction(0):
        min_boosted_fractional = (target_production - total_nodes_for_res * R_base_per_node) / denominator
    else:
        min_boosted_fractional = fractions.Fraction(0)

    min_boosted_fractional = max(fractions.Fraction(0), min_boosted_fractional)
    min_boosted_fractional = min(total_nodes_for_res, min_boosted_fractional)
    return min_boosted_fractional
```

```python
import numpy as np
from scipy.optimize import linprog
import fractions

# 5. Raw Resources List (consistency check)
raw_res_list = ['Wood Log', 'Coal', 'Iron Ore', 'Copper Ore', 'Stone', 'Wolframite', 'Uranium Ore']

# --- LP Scaling Factor (to mitigate floating-point inaccuracies) ---
LP_SCALE_FACTOR = fractions.Fraction(10**12) # Scale up problem coefficients for better float precision

# 1. Collect all unique items and map them to indices
all_items = set()
for recipe in recipes_library:
    all_items.add(recipe['produced_item'])
    for ing in recipe['ingredients']:
        all_items.add(ing['item'])

for res in raw_res_list:
    all_items.add(res)

all_items = sorted(list(all_items))
item_to_idx = {item: i for i, item in enumerate(all_items)}

# 2. Define LP variables: Each variable represents the execution rate of a single recipe
num_recipes = len(recipes_library)
recipe_names = [r['recipe_name'] for r in recipes_library]

# --- Fractional versions of LP components ---
c_fractional = np.zeros(num_recipes, dtype=object)
for i, recipe in enumerate(recipes_library):
    if recipe['produced_item'] == 'Earth Token':
        # Apply scale factor here for the objective function coefficients
        c_fractional[i] = -recipe['amount_produced'] * LP_SCALE_FACTOR

A_ub_base_fractional = []
for res in raw_res_list:
    row_fractional = np.zeros(num_recipes, dtype=object)
    for i, recipe in enumerate(recipes_library):
        for ing in recipe['ingredients']:
            if ing['item'] == res:
                # Apply scale factor here for A_ub matrix coefficients
                row_fractional[i] += ing['amount'] * LP_SCALE_FACTOR
    A_ub_base_fractional.append(row_fractional)

A_eq_fractional = []
b_eq_fractional = []

item_balance_matrix_fractional = np.zeros((len(all_items), num_recipes), dtype=object)

for recipe_idx, recipe in enumerate(recipes_library):
    produced_item_idx = item_to_idx[recipe['produced_item']]
    # Apply scale factor here for production components in A_eq matrix
    item_balance_matrix_fractional[produced_item_idx, recipe_idx] += recipe['amount_produced'] * LP_SCALE_FACTOR

    for ing in recipe['ingredients']:
        consumed_item_idx = item_to_idx[ing['item']]
        # Apply scale factor here for consumption components in A_eq matrix
        item_balance_matrix_fractional[consumed_item_idx, recipe_idx] -= ing['amount'] * LP_SCALE_FACTOR

for item_name in all_items:
    if item_name not in raw_res_list and item_name != 'Earth Token':
        row_fractional = item_balance_matrix_fractional[item_to_idx[item_name]]

        if item_name == 'Nuclear Fuel Cell':
            A_eq_fractional.append(row_fractional)
            # nuclear_fuel_cell_fixed_demand is already a fraction. Multiply by scale factor.
            b_eq_fractional.append(nuclear_fuel_cell_fixed_demand * LP_SCALE_FACTOR)
        else:
            A_eq_fractional.append(row_fractional)
            # Use fraction zero, multiply by scale factor to keep consistency
            b_eq_fractional.append(fractions.Fraction(0) * LP_SCALE_FACTOR)

# Convert to float arrays for the linprog call
c = np.array([float(x) for x in c_fractional])

A_ub = [] # This will be built for the initial run
b_ub = []
for i, res in enumerate(raw_res_list):
    row_float = np.array([float(x) for x in A_ub_base_fractional[i]])
    A_ub.append(row_float)
    # resource_production_rates has fractions. Multiply by scale factor.
    b_ub.append(float(resource_production_rates[res] * LP_SCALE_FACTOR))

A_eq = np.array([[float(x) for x in row] for row in A_eq_fractional])
b_eq = np.array([float(x) for x in b_eq_fractional])

# 6. Bounds: All recipe execution rates must be non-negative
bounds = [(0, None) for _ in range(num_recipes)]
```

```python
import pandas as pd
import collections # Import the collections module
import fractions # Import the fractions module
import math

def format_fraction_as_mixed(f_val):
    if f_val == 0: return "0"
    if f_val.denominator == 1: return str(f_val.numerator)

    sign = "-" if f_val < 0 else ""
    abs_f_val = abs(f_val)

    integer_part = abs_f_val.numerator // abs_f_val.denominator
    fraction_part = abs_f_val - integer_part

    if integer_part == 0:
        return f"{sign}{fraction_part.numerator}/{fraction_part.denominator}"
    elif fraction_part == 0:  # Exactly an integer
        return f"{sign}{integer_part}"
    else:  # Mixed number
        return f"{sign}{integer_part} and {fraction_part.numerator}/{fraction_part.denominator}"


# --- Pre-computation for Percentages and Recipe Options ---
item_total_produced_amount_optimal = collections.defaultdict(fractions.Fraction)
item_all_recipe_names = collections.defaultdict(list)

for recipe_data in recipes_library:
    item_all_recipe_names[recipe_data['produced_item']].append(recipe_data['recipe_name'])
    if recipe_data['recipe_name'] in optimal_recipe_rates:
        rate = optimal_recipe_rates[recipe_data['recipe_name']]
        item_total_produced_amount_optimal[recipe_data['produced_item']] += recipe_data['amount_produced'] * rate


# --- Store the final LP consumption as the "adjusted production" for reporting ---
final_lp_resource_production_rates_snapshot = final_raw_resource_consumption.copy()

# --- 1. Calculate Net Production/Consumption for all items ---
# Initialize net_flow for all possible items to zero
net_flow = {item: fractions.Fraction(0) for item in all_items}

for recipe_data in recipes_library:
    recipe_name = recipe_data['recipe_name']
    if recipe_name in optimal_recipe_rates:  # Only consider recipes active in the optimal solution
        rate = optimal_recipe_rates[recipe_name]  # This is already a fractions.Fraction

        # Add the produced item's amount to its net flow
        produced_item = recipe_data['produced_item']
        produced_amount = recipe_data['amount_produced'] * rate
        net_flow[produced_item] += produced_amount

        # Subtract each ingredient's amount from its net flow
        for ingredient in recipe_data['ingredients']:
            consumed_item = ingredient['item']
            consumed_amount = ingredient['amount'] * rate
            net_flow[consumed_item] -= consumed_amount

# --- 2. Prepare data for DataFrame with belt calculations ---
breakdown_data = []
belt_capacity = fractions.Fraction(480)  # A belt has a capacity of 480 items/minute

for item, net_amount_fraction in sorted(net_flow.items()):
    # Only include items that have a non-zero net production or consumption
    if net_amount_fraction != 0:
        net_amount_float = float(net_amount_fraction)

        belts_fraction = net_amount_fraction / belt_capacity
        belts_float = float(belts_fraction)

        breakdown_data.append({
            'Item': item,
            'Net Amount (Fraction)': format_fraction_as_mixed(net_amount_fraction),
            'Net Amount (Decimal)': net_amount_float,
            'Belts (Fraction)': format_fraction_as_mixed(belts_fraction),
            'Belts (Decimal)': belts_float
        })

# Create a Pandas DataFrame for the item flow and sort for readability
df_item_flow = pd.DataFrame(breakdown_data)
df_item_flow = df_item_flow.sort_values(by='Net Amount (Decimal)', ascending=False).reset_index(drop=True)


# --- Display Results ---

# --- Raw Resource Production (Adjusted to optimal LP Consumption) ---
raw_resource_breakdown = []
raw_resource_breakdown.append("\n--- Raw Resource Production (Adjusted to Consumption) ---")

for raw_res in raw_res_list:
    total_nodes_for_res = node_counts[raw_res] # From user input
    target_production_lp = final_lp_resource_production_rates_snapshot[raw_res] # Actual consumption from LP

    # Calculate R_base_per_node for the current resource
    if raw_res == uranium_resource:
        base_rate = uranium_default_base_rate if world_type == 'default' else uranium_gen2_base_rate
        R_base_per_node = base_rate # Uranium extractors do not level up, so no multiplier
    else:
        base_rate = default_world_base_rate if world_type == 'default' else gen2_world_base_rate
        # Apply extractor multiplier to standard resources
        R_base_per_node = base_rate * current_extractor_multiplier # Fix: Assign to R_base_per_node

    # Calculate total production from all nodes if NONE are boosted
    total_unboosted_production = total_nodes_for_res * R_base_per_node

    # Determine the minimum required nuclear boosts to meet LP consumption
    nuclear_boost_factor_current_world = boost_percentages['nuclear'][world_type]
    denominator = R_base_per_node * nuclear_boost_factor_current_world

    if denominator != fractions.Fraction(0):
        # Formula: N_boosted = (C - N_total * R_base) / (R_base * B_nuclear)
        min_boosted_fractional_for_display = (target_production_lp - total_nodes_for_res * R_base_per_node) / denominator
    else:
        min_boosted_fractional_for_display = fractions.Fraction(0) # Should not occur for resources with nuclear boost

    min_boosted_fractional_for_display = max(fractions.Fraction(0), min_boosted_fractional_for_display)
    min_boosted_fractional_for_display = min(total_nodes_for_res, min_boosted_fractional_for_display)


    # Determine the displayed production rate based on whether boosting is needed
    if min_boosted_fractional_for_display == fractions.Fraction(0):
        # If no boosting is required, display the full unboosted production capacity
        displayed_production_rate = total_unboosted_production
    else:
        # If boosting is required, display the LP target production rate
        displayed_production_rate = target_production_lp

    belts_produced = displayed_production_rate / belt_capacity

    raw_resource_breakdown.append(f"\n--- Resource '{raw_res}' ---")
    raw_resource_breakdown.append(f"  Adjusted Produced: {format_fraction_as_mixed(displayed_production_rate)} ({float(displayed_production_rate):.6f}) units/min")
    raw_resource_breakdown.append(f"  Belts for Adjusted Production: {format_fraction_as_mixed(belts_produced)} ({float(belts_produced):.6f}) belts")

    downstream_consumptions = []
    for recipe_data in recipes_library:
        recipe_name = recipe_data['recipe_name']
        if recipe_name in optimal_recipe_rates: # Only consider active recipes
            rate = optimal_recipe_rates[recipe_name]
            for ingredient in recipe_data['ingredients']:
                if ingredient['item'] == raw_res:
                    amount_consumed_by_this_recipe = ingredient['amount'] * rate
                    belts_for_consumption = amount_consumed_by_this_recipe / belt_capacity
                    downstream_consumptions.append(
                        f"    -> Used in '{recipe_name}': {format_fraction_as_mixed(amount_consumed_by_this_recipe)} ({float(amount_consumed_by_this_recipe):.6f}) units/min | Belts: {format_fraction_as_mixed(belts_for_consumption)} ({float(belts_for_consumption):.6f})")

    if downstream_consumptions:
        raw_resource_breakdown.append("  Downstream Consumption (by recipe contribution):")
        raw_resource_breakdown.extend(downstream_consumptions)
    else:
        raw_resource_breakdown.append("  No downstream consumption (raw resource may be surplus or not used).")

for line in raw_resource_breakdown:
    print(line)

# --- Minimum Nuclear-Boosted Nodes Calculation ---
min_boosted_nodes_breakdown = []
min_boosted_nodes_breakdown.append("\n--- Minimum Nuclear-Boosted Nodes Required (to match LP Consumption) ---")

for raw_res in raw_res_list:
    total_nodes_for_res = node_counts[raw_res] # From user input
    target_production = final_lp_resource_production_rates_snapshot[raw_res] # Actual consumption from LP

    # Calculate R_base_per_node for the current resource
    if raw_res == uranium_resource:
        base_rate = uranium_default_base_rate if world_type == 'default' else uranium_gen2_base_rate
        R_base_per_node = base_rate # Uranium extractors do not level up, so no multiplier
    else:
        base_rate = default_world_base_rate if world_type == 'default' else gen2_world_base_rate
        R_base_per_node = base_rate * current_extractor_multiplier

    nuclear_boost_factor_current_world = boost_percentages['nuclear'][world_type]

    # Calculate required boosted nodes
    denominator = R_base_per_node * nuclear_boost_factor_current_world
    if denominator != fractions.Fraction(0):
        # Formula: N_boosted = (C - N_total * R_base) / (R_base * B_nuclear)
        min_boosted_fractional = (target_production - total_nodes_for_res * R_base_per_node) / denominator
    else:
        min_boosted_fractional = fractions.Fraction(0) # Should not occur for resources with nuclear boost

    # Ensure min_boosted_fractional is not negative (if unboosted nodes produce enough or too much)
    min_boosted_fractional = max(fractions.Fraction(0), min_boosted_fractional)
    # Ensure min_boosted_fractional does not exceed the total available nodes
    min_boosted_fractional = min(total_nodes_for_res, min_boosted_fractional)

    min_boosted_nodes_breakdown.append(
        f"  {raw_res}: {format_fraction_as_mixed(min_boosted_fractional)} ({float(min_boosted_fractional):.6f}) out of {float(total_nodes_for_res):.0f} total nodes"
    )

for line in min_boosted_nodes_breakdown:
    print(line)


print("\n--- Detailed Item Flow (Net Production/Consumption per Minute) ---")
display(df_item_flow.to_string())

# --- 3. Generate Detailed Recipe Breakdown ---
detailed_recipe_breakdown = []

for recipe_data in recipes_library:
    recipe_name = recipe_data['recipe_name']
    if recipe_name in optimal_recipe_rates:  # Only process recipes that are actually used
        rate = optimal_recipe_rates[recipe_name]
        produced_item = recipe_data['produced_item']
        produced_per_batch = recipe_data['amount_produced']
        production_time_s = recipe_data['production_time_s']
        building_used = recipe_data['building_used']

        # --- Percentage of Use Calculation ---
        percentage_str = ""
        # Only apply percentage if the item has multiple recipe options in the library
        if len(item_all_recipe_names[produced_item]) > 1:
            total_output_for_this_item = item_total_produced_amount_optimal[produced_item]
            this_recipe_output_amount = produced_per_batch * rate
            if total_output_for_this_item > fractions.Fraction(0):
                percentage = (this_recipe_output_amount / total_output_for_this_item) * 100
                percentage_str = f" ({float(percentage):.6f}%)"
            else:  # Should not happen for an active recipe
                percentage_str = " (0.000000%)"

        # --- Product-Ingredient Ratio Display ---
        ratio_parts_summary = []

        if recipe_data['ingredients']:
            # Calculate per-unit amounts for scaling
            per_unit_product_amounts = []
            for ingredient in recipe_data['ingredients']:
                per_unit_product_amounts.append(ingredient['amount'] / produced_per_batch)

            # Find LCM of all denominators to get a scaling factor that makes all amounts integers
            scaling_factor_for_integers = fractions.Fraction(1)
            for f_val in per_unit_product_amounts:
                if f_val.denominator > 0:
                    common_multiple = (scaling_factor_for_integers.denominator * f_val.denominator) // math.gcd(scaling_factor_for_integers.denominator, f_val.denominator)
                    scaling_factor_for_integers = fractions.Fraction(scaling_factor_for_integers.numerator * (common_multiple // scaling_factor_for_integers.denominator), common_multiple)
                else:
                    scaling_factor_for_integers = fractions.Fraction(1)

            if scaling_factor_for_integers.denominator != 1:
                scaling_factor_for_integers = fractions.Fraction(1)

            display_produced_amount = fractions.Fraction(1) * scaling_factor_for_integers
            if display_produced_amount == 0 and produced_per_batch > 0:
                display_produced_amount = produced_per_batch
                scaling_factor_for_integers = produced_per_batch


            for ingredient in recipe_data['ingredients']:
                scaled_ing_amount = (ingredient['amount'] / produced_per_batch) * scaling_factor_for_integers
                ratio_parts_summary.append(f"{format_fraction_as_mixed(scaled_ing_amount)}  {ingredient['item']}")
            ratio_summary_str = f" | {format_fraction_as_mixed(display_produced_amount)}  {produced_item} : {' : '.join(ratio_parts_summary)} |"
        else:
            ratio_summary_str = f" | 1  {produced_item} : (no ingredients) |"

        detailed_recipe_breakdown.append(f"\n--- Recipe '{recipe_name}' ---{percentage_str}{ratio_summary_str}")

        total_produced_per_minute = produced_per_batch * rate
        belts_produced = total_produced_per_minute / belt_capacity

        # Calculate fractional buildings required
        items_per_minute_per_building_base = (produced_per_batch / production_time_s) * fractions.Fraction(60)
        assumed_tier_name = assumed_building_tiers.get(building_used, 4)  # Default to tier 4 (300%) if not specified
        tier_multiplier = building_tier_multipliers.get(assumed_tier_name, fractions.Fraction(1))  # Default to 1.0 if tier name not found
        items_per_minute_per_building_at_tier = items_per_minute_per_building_base * tier_multiplier

        # Avoid division by zero if a recipe produces 0 items/min
        if items_per_minute_per_building_at_tier > 0:
            fractional_buildings_required = total_produced_per_minute / items_per_minute_per_building_at_tier
        else:
            fractional_buildings_required = fractions.Fraction(0)  # Should not happen for active recipes

        detailed_recipe_breakdown.append(f"  Run Rate (batches/min): {format_fraction_as_mixed(rate)} ({float(rate):.6f})")
        detailed_recipe_breakdown.append(f"  Total Produced: {format_fraction_as_mixed(total_produced_per_minute)} ({float(total_produced_per_minute):.6f}) units/min of {produced_item}")
        detailed_recipe_breakdown.append(f"  Belts for Production: {format_fraction_as_mixed(belts_produced)} ({float(belts_produced):.6f}) belts")
        detailed_recipe_breakdown.append(f"  Buildings Required ({building_used} at {assumed_tier_name}): {format_fraction_as_mixed(fractional_buildings_required)} ({float(fractional_buildings_required):.6f})")

        # --- NEW SECTION: Downstream Consumption ---
        downstream_consumptions = []
        for consumer_recipe_data in recipes_library:
            consumer_recipe_name = consumer_recipe_data['recipe_name']
            if consumer_recipe_name in optimal_recipe_rates: # Only consider active consumer recipes
                consumer_rate = optimal_recipe_rates[consumer_recipe_name]
                for ingredient_in_consumer in consumer_recipe_data['ingredients']:
                    if ingredient_in_consumer['item'] == produced_item:
                        amount_consumed_by_this_recipe = ingredient_in_consumer['amount'] * consumer_rate
                        belts_for_consumption = amount_consumed_by_this_recipe / belt_capacity
                        downstream_consumptions.append(
                            f"    -> Used in '{consumer_recipe_name}': {format_fraction_as_mixed(amount_consumed_by_this_recipe)} ({float(amount_consumed_by_this_recipe):.6f}) units/min | Belts: {format_fraction_as_mixed(belts_for_consumption)} ({float(belts_for_consumption):.6f})")
        if downstream_consumptions:
            detailed_recipe_breakdown.append("  Downstream Consumption:")
            detailed_recipe_breakdown.extend(downstream_consumptions)

        if recipe_data['ingredients']:
            detailed_recipe_breakdown.append("  Ingredients Consumed:")
            for ingredient in recipe_data['ingredients']:
                ing_item = ingredient['item']
                ing_amount_per_batch = ingredient['amount']
                total_consumed_per_minute = ing_amount_per_batch * rate
                belts_consumed = total_consumed_per_minute / belt_capacity
                detailed_recipe_breakdown.append(f"    - {ing_item}: {format_fraction_as_mixed(total_consumed_per_minute)} ({float(total_consumed_per_minute):.6f}) units/min | Belts: {format_fraction_as_mixed(belts_consumed)} ({float(belts_consumed):.6f})) ")
        else:
            detailed_recipe_breakdown.append("  No ingredients required (raw resource extraction).")


print("\n--- Detailed Recipe Breakdown ---")
for line in detailed_recipe_breakdown:
    print(line)
```

```python
import numpy as np
from scipy.optimize import linprog
import fractions
import math
import collections # Required for defaultdict

def run_lp_with_num_plants(num_plants_for_lp):
    # Calculate fixed demand for Nuclear Fuel Cells based on the current num_plants_for_lp
    current_nuclear_fuel_cell_fixed_demand = num_plants_for_lp * fractions.Fraction(1, 2)

    # --- Re-initialize LP components that depend on nuclear plant count ---
    # These variables are assumed to be globally available from previous cells.
    # recipes_library, all_items, item_to_idx, num_recipes, LP_SCALE_FACTOR
    # c, A_ub, bounds, raw_res_list, resource_production_rates

    A_eq_fractional_local = []
    b_eq_fractional_local = []

    # Rebuild item balance matrix for equality constraints
    item_balance_matrix_fractional_local = np.zeros((len(all_items), num_recipes), dtype=object)

    for recipe_idx, recipe in enumerate(recipes_library):
        produced_item_idx = item_to_idx[recipe['produced_item']]
        item_balance_matrix_fractional_local[produced_item_idx, recipe_idx] += recipe['amount_produced'] * LP_SCALE_FACTOR

        for ing in recipe['ingredients']:
            consumed_item_idx = item_to_idx[ing['item']]
            item_balance_matrix_fractional_local[consumed_item_idx, recipe_idx] -= ing['amount'] * LP_SCALE_FACTOR

    for item_name in all_items:
        if item_name not in raw_res_list and item_name != 'Earth Token':
            row_fractional = item_balance_matrix_fractional_local[item_to_idx[item_name]]

            if item_name == 'Nuclear Fuel Cell':
                A_eq_fractional_local.append(row_fractional)
                b_eq_fractional_local.append(current_nuclear_fuel_cell_fixed_demand * LP_SCALE_FACTOR)
            else:
                A_eq_fractional_local.append(row_fractional)
                b_eq_fractional_local.append(fractions.Fraction(0) * LP_SCALE_FACTOR)

    # Convert to float arrays for the linprog call
    A_eq_local = np.array([[float(x) for x in row] for row in A_eq_fractional_local])
    b_eq_local = np.array([float(x) for x in b_eq_fractional_local])

    # b_ub needs to be recreated as resource_production_rates is a global variable.
    b_ub_local = []
    for res in raw_res_list:
        b_ub_local.append(float(resource_production_rates[res] * LP_SCALE_FACTOR))

    # Solve the LP problem
    res_lp = linprog(c, A_ub=A_ub, b_ub=b_ub_local, A_eq=A_eq_local, b_eq=b_eq_local, bounds=bounds, method='highs')

    if res_lp.success:
        max_earth_tokens = -res_lp.fun / LP_SCALE_FACTOR
        optimal_recipe_rates = {}
        for i, rate in enumerate(res_lp.x):
            if rate > 1e-9:  # Only show recipes with a significant execution rate
                recipe_name = recipe_names[i]
                optimal_recipe_rates[recipe_name] = fractions.Fraction(str(rate))

        # Recalculate raw resource consumption based on the new optimal rates
        final_raw_resource_consumption = {res: fractions.Fraction(0) for res in raw_res_list}
        for recipe_idx, recipe in enumerate(recipes_library):
            if res_lp.x[recipe_idx] > 1e-9:
                execution_rate = fractions.Fraction(str(res_lp.x[recipe_idx]))
                for ing in recipe['ingredients']:
                    if ing['item'] in raw_res_list:
                        final_raw_resource_consumption[ing['item']] += ing['amount'] * execution_rate
        return max_earth_tokens, optimal_recipe_rates, final_raw_resource_consumption
    else:
        print(f"LP FAILED with {float(num_plants_for_lp):.6f} nuclear plants: {res_lp.message}")
        return None, None, None
```

```python
import math
import fractions

def solve_lp_iteratively(
    initial_effective_plants,
    original_num_plants, # e.g., 50 (initial integer count of nuclear plants)
    wood_log_covered_by_previous_plants, # e.g., 149 (user-defined fixed allocation)
    wood_log_covered_by_final_plant, # e.g., 7 (user-defined fixed allocation)
    resource_to_monitor='Wood Log',
    tolerance=fractions.Fraction(1, 10**10), # Using fractions for precision in tolerance
    max_iterations=20
):
    print(f"Starting iterative LP solver for '{resource_to_monitor}' with initial effective plants: {float(initial_effective_plants):.10f}")

    # --- Iteration 0 (Initial LP Run with Original Plants) ---
    print("\n--- Iteration 0 (Initial LP Run with Original Plants) ---")
    print(f"Original num_nuclear_plants used in initial LP: {float(original_num_plants):.0f}")

    # Run LP with the original number of plants to get the initial resource consumption
    _, _, final_raw_resource_consumption_initial_lp = run_lp_with_num_plants(original_num_plants)

    if final_raw_resource_consumption_initial_lp is None:
        print("Initial LP run failed. Aborting.")
        return initial_effective_plants, None, None, ["Initial LP run failed."]

    # Calculate min_boosted_fractional_for_res for Wood Log based on initial LP run
    total_nodes_for_res_initial = node_counts[resource_to_monitor]
    target_production_initial = final_raw_resource_consumption_initial_lp[resource_to_monitor]

    min_boosted_fractional_for_res_initial = calculate_min_boosted_fractional(
        target_production=target_production_initial,
        total_nodes_for_res=total_nodes_for_res_initial,
        resource_name=resource_to_monitor,
        world_type_val=world_type,
        extractor_multiplier_val=current_extractor_multiplier
    )

    print(f"Initial Min boosted fractional for '{resource_to_monitor}': {format_fraction_as_mixed(min_boosted_fractional_for_res_initial)} ({float(min_boosted_fractional_for_res_initial):.6f})")

    # Calculate quotient for Iteration 0
    boost_needed_from_final_plant_equivalent_initial = min_boosted_fractional_for_res_initial - wood_log_covered_by_previous_plants
    boost_needed_from_final_plant_equivalent_initial = max(fractions.Fraction(0), boost_needed_from_final_plant_equivalent_initial)

    if wood_log_covered_by_final_plant > fractions.Fraction(0):
        quotient_initial = boost_needed_from_final_plant_equivalent_initial / wood_log_covered_by_final_plant
    else:
        quotient_initial = fractions.Fraction(0)

    new_effective_plants_initial = (original_num_plants - 1) + quotient_initial

    print(f"  wood_log_covered_by_previous_plants: {format_fraction_as_mixed(wood_log_covered_by_previous_plants)} ({float(wood_log_covered_by_previous_plants):.6f})")
    print(f"  wood_log_covered_by_final_plant: {format_fraction_as_mixed(wood_log_covered_by_final_plant)} ({float(wood_log_covered_by_final_plant):.6f})")
    print(f"  Numerator for quotient: {format_fraction_as_mixed(boost_needed_from_final_plant_equivalent_initial)} ({float(boost_needed_from_final_plant_equivalent_initial):.6f})")
    print(f"  Denominator for quotient: {format_fraction_as_mixed(wood_log_covered_by_final_plant)} ({float(wood_log_covered_by_final_plant):.6f})")
    print(f"  Calculated quotient for Iteration 0: {format_fraction_as_mixed(quotient_initial)} ({float(quotient_initial):.6f})")
    print(f"  New effective total plants for Iteration 0: {format_fraction_as_mixed(new_effective_plants_initial)} ({float(new_effective_plants_initial):.10f})")

    warnings = [] # Initialize warnings list

    if quotient_initial > fractions.Fraction(1):
        warnings.append(f"WARNING in Iteration 0: Calculated quotient ({float(quotient_initial):.6f}) > 1. This means the 'final plant' allocation ({float(wood_log_covered_by_final_plant):.0f}) is insufficient to cover the required boost ({float(boost_needed_from_final_plant_equivalent_initial):.6f}).")



    current_effective_plants = initial_effective_plants
    previous_effective_plants = fractions.Fraction(-1) # To ensure loop runs at least once
    iteration = 0

    convergence_achieved = False

    while iteration < max_iterations and abs(current_effective_plants - previous_effective_plants) > tolerance:
        iteration += 1
        print(f"\n--- Iteration {iteration} ---")
        previous_effective_plants = current_effective_plants

        # 1. Run the LP with the current effective number of nuclear plants
        max_earth_tokens, optimal_recipe_rates, final_raw_resource_consumption = run_lp_with_num_plants(current_effective_plants)

        if max_earth_tokens is None:
            print(f"LP failed at iteration {iteration}. Aborting.")
            break

        print(f"Max Earth Tokens: {float(max_earth_tokens):.6f}")

        # 2. Calculate min_boosted_fractional for the monitored resource
        total_nodes_for_res = node_counts[resource_to_monitor]
        target_production = final_raw_resource_consumption[resource_to_monitor]

        min_boosted_fractional_for_res = calculate_min_boosted_fractional(
            target_production=target_production,
            total_nodes_for_res=total_nodes_for_res,
            resource_name=resource_to_monitor,
            world_type_val=world_type,
            extractor_multiplier_val=current_extractor_multiplier
        )

        print(f"Min boosted fractional for '{resource_to_monitor}': {format_fraction_as_mixed(min_boosted_fractional_for_res)} ({float(min_boosted_fractional_for_res):.6f})")

        # 3. Recalculate new_effective_total_plants based on user's fixed allocations
        boost_needed_from_final_plant_equivalent = min_boosted_fractional_for_res - wood_log_covered_by_previous_plants
        boost_needed_from_final_plant_equivalent = max(fractions.Fraction(0), boost_needed_from_final_plant_equivalent)

        # Debug prints for quotient calculation
        print(f"  wood_log_covered_by_previous_plants: {format_fraction_as_mixed(wood_log_covered_by_previous_plants)} ({float(wood_log_covered_by_previous_plants):.6f})")
        print(f"  wood_log_covered_by_final_plant: {format_fraction_as_mixed(wood_log_covered_by_final_plant)} ({float(wood_log_covered_by_final_plant):.6f})")
        print(f"  Numerator for quotient: {format_fraction_as_mixed(boost_needed_from_final_plant_equivalent)} ({float(boost_needed_from_final_plant_equivalent):.6f})")
        print(f"  Denominator for quotient: {format_fraction_as_mixed(wood_log_covered_by_final_plant)} ({float(wood_log_covered_by_final_plant):.6f})")

        if wood_log_covered_by_final_plant > fractions.Fraction(0):
            quotient = boost_needed_from_final_plant_equivalent / wood_log_covered_by_final_plant
        else:
            quotient = fractions.Fraction(0)

        new_effective_plants = (original_num_plants - 1) + quotient
        print(f"  Calculated quotient: {format_fraction_as_mixed(quotient)} ({float(quotient):.6f})")
        print(f"Recalculated effective total plants: {format_fraction_as_mixed(new_effective_plants)} ({float(new_effective_plants):.10f})")

        # Add warning if quotient > 1
        if quotient > fractions.Fraction(1):
            warnings.append(f"WARNING in Iteration {iteration}: Calculated quotient ({float(quotient):.6f}) > 1. This means the 'final plant' allocation ({float(wood_log_covered_by_final_plant):.0f}) is insufficient to cover the required boost ({float(boost_needed_from_final_plant_equivalent):.6f}).")

        # Check for integer boundary changes in effective_total_plants
        old_integer_part_current_iter = math.floor(float(previous_effective_plants))
        new_integer_part_current_iter = math.floor(float(new_effective_plants))

        # Convert original_num_plants to integer for comparison
        original_num_plants_int = int(original_num_plants)

        # The warning should only be triggered if the new effective integer part is
        # greater than the original number of plants, or less than (original_num_plants - 1).
        # This allows for solutions like 24.x when original_num_plants is 25,
        # which means the solution is effectively using one less plant.
        if new_integer_part_current_iter > original_num_plants_int or new_integer_part_current_iter < (original_num_plants_int - 1):
            warnings.append(f"WARNING in Iteration {iteration}: The effective integer number of nuclear plants changed from {old_integer_part_current_iter} to {new_integer_part_current_iter}. This suggests a significant shift in resource requirements or the assumed allocation may need re-evaluation.")

        # Check if min_boosted_fractional for Wood Log is within user's specified bounds
        user_total_covered_nodes = wood_log_covered_by_previous_plants + wood_log_covered_by_final_plant

        if min_boosted_fractional_for_res > user_total_covered_nodes:
            warnings.append(f"WARNING in Iteration {iteration}: Required boosted '{resource_to_monitor}' nodes ({float(min_boosted_fractional_for_res):.6f}) exceed the user's total allocated capacity ({float(user_total_covered_nodes):.6f}). User input may be needed.")
        elif min_boosted_fractional_for_res < wood_log_covered_by_previous_plants:
             warnings.append(f"WARNING in Iteration {iteration}: Required boosted '{resource_to_monitor}' nodes ({float(min_boosted_fractional_for_res):.6f}) fall below the nodes covered by previous plants ({float(wood_log_covered_by_previous_plants):.6f}). User input may be needed.")

        current_effective_plants = new_effective_plants

        if abs(current_effective_plants - previous_effective_plants) <= tolerance:
            convergence_achieved = True
            print(f"\nConvergence achieved after {iteration} iterations. Difference: {format_fraction_as_mixed(abs(current_effective_plants - previous_effective_plants))} ({float(abs(current_effective_plants - previous_effective_plants)):.10f})")
            break

    if not convergence_achieved:
        print(f"\nMax iterations ({max_iterations}) reached without convergence. Final difference: {format_fraction_as_mixed(abs(current_effective_plants - previous_effective_plants))} ({float(abs(current_effective_plants - previous_effective_plants)):.10f})")

    print("\n--- Final Iteration Results ---")
    print(f"Converged Effective Total Plants: {format_fraction_as_mixed(current_effective_plants)} ({float(current_effective_plants):.10f})")
    print(f"Final Max Earth Tokens: {float(max_earth_tokens):.6f}")
    print(f"Final Min Boosted Fractional for '{resource_to_monitor}': {format_fraction_as_mixed(min_boosted_fractional_for_res)} ({float(min_boosted_fractional_for_res):.6f})")

    if warnings:
        print("\n--- Warnings ---")
        for w in warnings:
            print(w)

    return current_effective_plants, max_earth_tokens, min_boosted_fractional_for_res, warnings
```
